```python
import spacy
nlp = spacy.load("pt_core_news_lg")
texto_1 = open("./dados/Harry Potter e A Pedra Filosofal.txt", "r", encoding="utf-8")
texto_2 = open("./dados/Harry_Potter_Camara_Secreta-br.txt","r",encoding="utf-8")
texto_1 = texto_1.read()
texto_2 = texto_2.read()
doc_1 = nlp(texto_1)
doc_2 = nlp(texto_2)
```

```python
sentences = []

for sent_1 in doc_1.sents:
    sent= []
    for palavra in sent_1:
        if not palavra.is_punct and not palavra.is_space and not palavra.is_stop:
            sent.append(palavra.text.lower()) 
    if sent:
        sentences.append(sent)

for sent_2 in doc_2.sents:
    sent= []
    for palavra in sent_2:
        if not palavra.is_punct and not palavra.is_space and not palavra.is_stop:
            sent.append(palavra.text.lower())     
    if sent:
        sentences.append(sent)

print(f"Total de sentenças limpas extraídas: {len(sentences)}")
#for sent in sentences:
#    print(sent)
```

```python
from gensim.models import Word2Vec

model = Word2Vec(sentences, vector_size=100, window=5, min_count=5, sg=0, epochs=20, workers=3)
```

## Visualização do modelo

```python
model.wv.save_word2vec_format('model_harry.txt', binary=False)
#http://projector.tensorflow.org/
```
